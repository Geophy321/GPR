```python
from readgssi import readgssi
import matplotlib.pyplot as plt
import numpy as np
```

```python
from numpy import loadtxt, meshgrid,linspace
from matplotlib import pyplot as plt
from scipy.interpolate import griddata

x,t,a = loadtxt("ASCII/COHERENCY.ASC", usecols=(0,1,2),unpack=1)
a /= max(a) #normalizar perfil
fig = plt.figure()
ax = fig.add_subplot(111)
ax.grid(c="lightblue",lw=.5)
ax.set_xlabel("Distance (m)")
ax.set_ylabel("Time(ns) (ns)")
ax.set_title("COHERENCY.ASC")
ax.invert_yaxis()
N = 200
x_grid = linspace(x.min(), x.max(), N)
y_grid = linspace(t.min(), t.max(), N)
xi,yi = meshgrid(x_grid,y_grid)
zi = griddata((x, t), a,(xi,yi), method = 'linear')
img = ax.imshow(zi, cmap='gray', interpolation='bicubic',aspect="auto")                  
cb = fig.colorbar(img,label="amplitude")
plt.show()
```
